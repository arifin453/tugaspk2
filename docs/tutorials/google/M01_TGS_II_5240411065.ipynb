```
# Soal 1: Program Menghitung Usia
from datetime import datetime

# Input dari pengguna
nama = input("Masukkan Nama Lengkap: ")
tanggal_lahir = int(input("Masukkan Tanggal Lahir (1-31): "))
bulan_lahir = int(input("Masukkan Bulan Lahir (1-12): "))
tahun_lahir = int(input("Masukkan Tahun Lahir (YYYY): "))

# Hitung usia
hari_ini = datetime(2025, 2, 12)
tanggal_lahir_user = datetime(tahun_lahir, bulan_lahir, tanggal_lahir)
usia = hari_ini.year - tanggal_lahir_user.year
if (hari_ini.month, hari_ini.day) < (tanggal_lahir_user.month, tanggal_lahir_user.day):
    usia -= 1

# Output
print(f"Hai, {nama}! Pada hari ini Rabu 12 Februari 2025 Anda berumur {usia} tahun.")
print("Selamat belajar Struktur Data!")
```

```
# Soal 2: Operasi Matematika pada Dua Bilangan Bulat
# Input dari pengguna
bilangan1 = int(input("Masukkan bilangan pertama: "))
bilangan2 = int(input("Masukkan bilangan kedua: "))

# Operasi matematika
penjumlahan = bilangan1 + bilangan2
pengurangan = bilangan1 - bilangan2
perkalian = bilangan1 * bilangan2
pembagian = bilangan1 / bilangan2 if bilangan2 != 0 else "Tidak dapat dibagi dengan nol"
sisa_bagi = bilangan1 % bilangan2 if bilangan2 != 0 else "Tidak ada sisa karena pembagian nol"

# Output hasil
print("\nHasil Perhitungan:")
print(f"Penjumlahan: {bilangan1} + {bilangan2} = {penjumlahan}")
print(f"Pengurangan: {bilangan1} - {bilangan2} = {pengurangan}")
print(f"Perkalian: {bilangan1} * {bilangan2} = {perkalian}")
print(f"Pembagian: {bilangan1} / {bilangan2} = {pembagian}")
if isinstance(sisa_bagi, int):
    print(f"Sisa Pembagian: {bilangan1} % {bilangan2} = {sisa_bagi}")
```
